# 1. OPTIONS — Contracts That Give You Choices
---

## 1.1 What Is an Option?

An option is a **contract that gives you the right, but not the obligation**, to buy or sell an asset at a predetermined price before a certain date.

### The Two Types

| Type | Right | Buyer hopes | Seller hopes |
|------|-------|-------------|---------------|
| **Call** | Right to BUY at strike price | Price goes UP | Price stays below strike |
| **Put** | Right to SELL at strike price | Price goes DOWN | Price stays above strike |

### The Key Asymmetry

This is what makes options fundamentally different from stocks or futures:

- **Buyer**: Pays a premium upfront. Maximum loss = premium paid. Potential gain = unlimited (calls) or large (puts). The buyer has a **convex** payoff — they benefit disproportionately from large moves.

- **Seller**: Receives the premium. Maximum gain = premium received. Potential loss = unlimited (calls) or large (puts). The seller has a **concave** payoff — they collect small, steady income but face catastrophic risk.

This buyer-seller asymmetry is why options exist: the buyer is purchasing **insurance against adverse moves**, and the seller is the insurance company.

### Moneyness

| Term | Call | Put | Intuition |
|------|------|-----|----------|
| In-the-money (ITM) | S > K | S < K | Exercising now would be profitable |
| At-the-money (ATM) | S ≈ K | S ≈ K | Right at the boundary |
| Out-of-the-money (OTM) | S < K | S > K | Exercising now would be a loss — you wouldn't do it |

### Intrinsic vs Time Value

$$\text{Option Price} = \text{Intrinsic Value} + \text{Time Value}$$

- **Intrinsic value**: What the option is worth if exercised RIGHT NOW. For a call: $\max(S - K, 0)$.
- **Time value**: The extra premium you pay for the POSSIBILITY that the option becomes more valuable before expiry. Time value decays as expiration approaches (theta decay).

An OTM option has zero intrinsic value — its entire price is time value (hope).

## 1.2 The Black-Scholes Model — A Revolution in Finance

### The Big Question

Before Black-Scholes (1973), nobody agreed on how to price options. Is a call on IBM worth $5 or $15? It was pure guesswork.

Fischer Black, Myron Scholes, and Robert Merton showed that under certain assumptions, there is exactly **one correct price** for an option, and you can derive it mathematically.

### The Key Insight: Hedging Eliminates Risk

The genius of Black-Scholes is NOT the formula itself — it's the **argument** behind it:

1. You sell a call option.
2. You buy exactly $\Delta$ shares of the stock (delta hedging).
3. This combination — short option + long stock — has **zero net sensitivity** to small stock moves.
4. A riskless portfolio must earn the risk-free rate (no arbitrage).
5. This constraint uniquely determines the option price.

### Risk-Neutral Pricing — The Deepest Idea

Because you can hedge away all the stock risk, **the expected return of the stock doesn't matter** for option pricing. You don't need to know whether the stock will go up 10% or 30% per year.

Instead, you can price options AS IF everyone were risk-neutral (i.e., as if the stock grew at the risk-free rate). This simplifies everything enormously.

This is not a claim about investor psychology — it's a mathematical consequence of no-arbitrage hedging.

### The Five Inputs

| Input | Observable? | Sensitivity |
|-------|-------------|-------------|
| Stock price ($S$) | Yes — market data | High |
| Strike price ($K$) | Yes — contract terms | High |
| Time to expiry ($T$) | Yes — calendar | Medium |
| Risk-free rate ($r$) | Yes — Treasury yields | Low |
| Volatility ($\sigma$) | **NO — must estimate** | **Very high** |

Four of the five inputs are directly observable. Only volatility must be estimated or implied from market prices. This is why volatility is the **central variable** in options trading.

### What Black-Scholes Gets Wrong

The model assumes:
1. **Constant volatility** → In reality, volatility changes constantly (stochastic vol)
2. **Log-normal returns** → In reality, returns have fat tails and skewness
3. **Continuous trading** → In reality, markets close, have gaps, and can freeze
4. **No transaction costs** → In reality, every hedge costs money
5. **No dividends** → Easy to fix (adjusted Black-Scholes)
6. **European exercise only** → American options can be exercised early

Despite all these flaws, Black-Scholes is still the **universal language** of options markets. Everyone knows the model is wrong, but everyone uses it as a common reference point (like quoting distance in miles even though the Earth is curved).

# 2. THE GREEKS — Measuring Option Sensitivities
---

## 2.1 Why Greeks Matter

An option's price depends on five inputs. When any of them changes, the option price changes too. The Greeks measure **how much and in what way**.

Think of it like driving a car:
- **Delta** = speed (how fast you're moving)
- **Gamma** = acceleration (how fast your speed is changing)
- **Theta** = fuel consumption (you're burning money as time passes)
- **Vega** = how slippery the road is (sensitivity to uncertainty)
- **Rho** = effect of headwinds (interest rate sensitivity)

A professional options trader doesn't think in terms of option prices — they think in terms of Greeks.

## 2.2 Delta ($\Delta$) — The Hedge Ratio

### What It Means

Delta tells you: **"If the stock moves $1, how much does my option move?"**

- Call delta ranges from 0 to +1
- Put delta ranges from -1 to 0

### Three Ways to Think About Delta

#### 1. Price Sensitivity
A call with delta = 0.60 gains about $0.60 when the stock goes up $1, and loses about $0.60 when the stock drops $1. It behaves like owning 60% of a share.

#### 2. Hedge Ratio
To hedge a short call position (you sold 100 calls), buy $\Delta \times 100$ shares. If delta = 0.60, buy 60 shares. Now small stock moves don't affect your P&L.

#### 3. Probability Proxy
Delta ≈ probability the option expires in-the-money. A delta-0.30 call has roughly a 30% chance of finishing ITM. This is approximate but extremely useful for quick intuition.

### Delta Across Moneyness

| Moneyness | Call Delta | Behavior |
|-----------|-----------|----------|
| Deep ITM | ~1.0 | Moves like the stock itself |
| ATM | ~0.5 | Moves at half the stock's rate |
| Deep OTM | ~0.0 | Barely reacts to stock moves |

### Why Delta Changes

Delta is not constant — it moves as the stock price changes, time passes, and volatility shifts. This is why you need gamma.

## 2.3 Gamma ($\Gamma$) — The Curvature

### What It Means

Gamma tells you: **"How fast is delta changing?"**

It's the second derivative — the "acceleration" of the option price.

### Why Gamma Matters

If delta is your hedge, gamma tells you **how quickly your hedge goes stale**.

- **High gamma**: Delta changes rapidly with the stock price. You need to re-hedge frequently. Hedging is expensive and error-prone.
- **Low gamma**: Delta is stable. Your hedge stays accurate for a while. Less work, less cost.

### When Gamma Is Highest

Gamma is highest for **ATM options near expiration**. This makes intuitive sense:

Imagine a call option that expires tomorrow with strike $100. The stock is at $100 (ATM).
- If the stock goes to $101: the call is almost certainly ITM → delta jumps to ~1.0
- If the stock drops to $99: the call is almost certainly OTM → delta drops to ~0.0

A tiny stock move causes a massive delta swing. That's extreme gamma.

### Gamma and P&L

- **Long gamma** (own options): You BENEFIT from large moves in either direction. Each move makes your delta more favorable. But you pay theta (time decay) for this privilege.
- **Short gamma** (sold options): You're HURT by large moves. Each move makes your hedge more wrong. But you collect theta as compensation.

The theta-gamma tradeoff is the fundamental tension in options trading:

$$\text{Long gamma} \leftrightarrow \text{Pay theta}$$
$$\text{Short gamma} \leftrightarrow \text{Earn theta}$$

### The Gamma Squeeze

When market makers are massively short gamma (common in heavily traded meme stocks), a price move forces them to hedge by buying more stock (for calls), which pushes the price further, which increases delta further, forcing more buying. This positive feedback loop is a **gamma squeeze** — it amplifies moves far beyond fundamentals.

## 2.4 Theta ($\Theta$) — Time Decay

### What It Means

Theta tells you: **"How much value does my option lose each day, just from time passing?"**

All else being equal, options lose value every day. This is because less time = less opportunity for a favorable move.

### The Clock Is Always Ticking

- For the option **buyer**: Theta is the enemy. Your option is a melting ice cube. Every day it's worth less.
- For the option **seller**: Theta is your friend. You collect value as time passes. You're earning rent on the insurance you sold.

### Non-Linear Decay

Time decay is NOT constant — it **accelerates** as expiration approaches.

An ATM option with 90 days left might lose $0.05/day. With 10 days left, it might lose $0.20/day. In the final 2-3 days, decay is brutal.

This is why short-term option selling is popular: you earn the most theta in the final days.

### Weekend and Holiday Theta

A controversial topic: does theta tick on weekends? Technically yes (calendar time passes), but the stock can't move (no trading). The market handles this ambiguously — some pricing models use calendar days, others use trading days.

## 2.5 Vega ($\nu$) — Volatility Sensitivity

### What It Means

Vega tells you: **"If implied volatility increases by 1 percentage point, how much does my option price change?"**

### Why Vega Is King

Professional options traders often care more about vega than delta. Here's why:

Delta risk can be hedged easily (buy/sell the underlying stock). But vega risk — the risk that volatility itself changes — is much harder to hedge. You'd need to trade other options against your position.

Many options trades are fundamentally **volatility bets**:
- **Buying a straddle** (call + put at same strike): You're long vega — you profit if vol rises.
- **Selling an iron condor**: You're short vega — you profit if vol falls.

### Vega Across the Surface

- **ATM options**: Highest vega (most sensitive to vol changes)
- **Deep ITM/OTM options**: Low vega (vol barely matters because outcome is already ~certain)
- **Long-dated options**: Higher vega (more time for vol to play out)
- **Short-dated options**: Lower vega (not enough time for vol changes to matter)

### Vol-of-Vol: The Second Layer

Just as gamma measures how delta changes, **vanna** and **volga** measure how vega changes:
- **Vanna**: How does vega change when the stock moves? (Cross-effect of stock and vol)
- **Volga** (vomma): How does vega change when vol changes? (Convexity in vol)

These second-order Greeks matter for exotic options and for market makers managing large portfolios.

# 3. STOCHASTIC PROCESSES — Modeling Randomness
---

## 3.1 Why We Need Stochastic Calculus

### The Problem

Financial prices move randomly. We need mathematics that can handle randomness in continuous time. Regular calculus (Newton, Leibniz) was designed for smooth, deterministic functions. It breaks down when the function is random and jagged.

### What's Different About Stochastic Calculus?

In regular calculus:
- Functions are smooth → derivatives exist
- $(dx)^2 \approx 0$ (negligibly small, can be ignored)
- Chain rule: $df = f'(x) \, dx$

In stochastic calculus:
- Paths are jagged → no traditional derivative exists
- $(dW)^2 = dt$ (NOT negligible! This is the critical difference)
- Ito's chain rule: $df = f'(x) \, dx + \frac{1}{2} f''(x) \sigma^2 \, dt$ (extra correction term)

That extra term changes everything. It's why option prices have a term proportional to gamma (second derivative). It's why log-normal stock prices drift differently from what you'd naively expect.

## 3.2 Brownian Motion — The Foundation

### What Is It?

Brownian motion (named after Robert Brown, who observed pollen grains jittering in water) is a **continuous random path** that starts at zero.

### Physical Intuition

Imagine a drunk person starting at a lamppost, taking random steps:
- Each step is independent of the previous ones.
- Steps are normally distributed (small steps are more common than large ones).
- The path is continuous (no teleportation).
- After time $t$, the person's position is $N(0, t)$ — average position is still the lamppost, but uncertainty grows with time.

### Paradoxical Properties

Brownian motion has beautiful but counterintuitive properties:

1. **Continuous everywhere, differentiable nowhere**: The path has no smooth segments — it changes direction infinitely often in any interval, no matter how small. It's the ultimate jagged line.

2. **Infinite variation, finite quadratic variation**: If you measure total up-and-down movement (total variation), it's infinite. But if you sum squared increments (quadratic variation), it equals exactly $T$. This finite QV is what makes Ito calculus work.

3. **Self-similar**: Zoom in on any segment and it looks statistically identical to the whole path (fractal-like).

4. **Recurrence in 1D**: A 1D Brownian motion returns to zero infinitely often (but the waiting time between returns grows).

### Why Finance Uses It

Brownian motion is the **mathematical engine of randomness** in continuous-time finance. Every model — Black-Scholes, Heston, Hull-White — is built on top of Brownian motion. It's the equivalent of the normal distribution for continuous-time models.

## 3.3 Geometric Brownian Motion — How Stocks Are Modeled

### The Problem with Plain Brownian Motion

Raw Brownian motion ($W_t$) can go negative. Stock prices can't go negative (at worst, a stock goes to zero). Also, a $1 move means more to a $10 stock than a $1000 stock.

### The Solution: Multiplicative Noise

Instead of adding a random increment to the price, we add a random **percentage** change:

$$\frac{dS}{S} = \mu \, dt + \sigma \, dW$$

This says: the **percentage change** in the stock has a drift part ($\mu \, dt$) and a random part ($\sigma \, dW$).

### Key Consequences

1. **Stock prices are log-normally distributed**: $\ln(S_T) \sim N(\ln(S_0) + (\mu - \sigma^2/2)T, \sigma^2 T)$

2. **Prices can't go negative**: The exponential function ensures $S > 0$ always.

3. **The $\sigma^2/2$ correction**: Due to Ito's Lemma, the expected log-return is $\mu - \sigma^2/2$, not $\mu$. This is the "volatility drag" — higher volatility reduces the geometric average return.

4. **Multiplicative returns**: A 10% gain followed by a 10% loss doesn't return you to the starting point. GBM naturally captures this compounding effect.

### Limitations in Practice

GBM is a reasonable first approximation but misses:
- **Volatility clustering**: Calm and turbulent periods alternate (use GARCH or stochastic vol models)
- **Jumps**: Sudden large moves from news events (use jump-diffusion models)
- **Fat tails**: Extreme returns are more common than log-normal suggests
- **Mean reversion**: Some assets (interest rates, volatility, commodity spreads) don't random-walk — they revert to a long-term level

## 3.4 Ito's Lemma — The Chain Rule for Random Worlds

### The Problem

You know how $S$ (stock price) evolves. Now you want to know how $f(S)$ evolves — for example, how does $\ln(S)$ change? Or $S^2$? Or the price of an option $C(S, t)$?

In regular calculus, you'd use the chain rule: $df = f'(S) \, dS$. Done.

In stochastic calculus, that's **wrong** because the squared term $(dS)^2$ is not negligible.

### The Fix

Ito's Lemma adds a correction term that accounts for the roughness of Brownian motion. It's like saying: the chain rule works, BUT there's a bonus term proportional to the curvature of $f$ times the variance of the noise.

### Why the Correction Exists — An Analogy

Imagine driving on a curvy road:
- On a **straight road** ($f'' = 0$), small steering wheel jitters don't matter — you stay on course.
- On a **curved road** ($f'' \neq 0$), the same jitters cause you to systematically drift to one side. The tighter the curve, the more you drift.

The Ito correction captures this systematic drift caused by the interaction of randomness and curvature.

### The Most Important Application

Applying Ito's Lemma to $f(S) = \ln(S)$ under GBM gives:

$$d(\ln S) = (\mu - \sigma^2/2) \, dt + \sigma \, dW$$

The $-\sigma^2/2$ term is purely from Ito's correction. It means:
- The arithmetic average return is $\mu$
- The geometric average return is $\mu - \sigma^2/2$
- Higher volatility → lower compound growth rate

This is the mathematical origin of **volatility drag**.

## 3.5 Mean Reversion — Things That Come Back

### The Concept

Not everything in finance random-walks. Some quantities have a natural "equilibrium" or "normal level" that they tend to return to after being pushed away.

### Examples of Mean-Reverting Quantities

| Quantity | Mean-reverts to | Why |
|----------|----------------|------|
| Interest rates | Central bank target rate | Central bank policy actively pulls rates back |
| Volatility (VIX) | ~15-20 | Panic subsides, hedging demand normalizes |
| Pairs spread (Coca-Cola vs Pepsi) | Historical average spread | Fundamental similarity limits divergence |
| Commodity prices | Cost of production | Prices below cost → production cuts → supply shrinks → price rises |
| Exchange rates | Purchasing power parity (very long-term) | Goods become cheap/expensive → trade flows adjust |

### Random Walk vs Mean Reversion vs Momentum

| Behavior | Today's move predicts tomorrow's? | Trading strategy |
|----------|-----------------------------------|-------------------|
| Random walk | No — moves are independent | Can't be profitably traded |
| Mean reversion | Yes — in the opposite direction | Buy dips, sell rallies |
| Momentum | Yes — in the same direction | Buy winners, sell losers |

### The Ornstein-Uhlenbeck Process

The standard mathematical model for mean reversion. Think of it as a random walk with a **rubber band** attached to a fixed point:

- The further you wander from home, the **harder the rubber band pulls you back**.
- But random shocks keep pushing you around.
- The result: you oscillate around the mean, never staying far away for long.

### Half-Life

The **half-life** of mean reversion is the time it takes for a deviation to shrink by half. It's a key parameter for pairs traders:

- **Short half-life** (hours to days): Good for high-frequency pairs trading.
- **Long half-life** (months to years): Too slow for most trading strategies — by the time it reverts, you've paid too much in carry costs.

### Mean Reversion in Practice: Pairs Trading

1. Find two historically correlated stocks (e.g. Visa and Mastercard).
2. Compute the spread (price ratio or difference).
3. When the spread widens (one stock diverges): short the outperformer, long the underperformer.
4. When the spread reverts to normal: close both positions for a profit.

Risk: sometimes spreads diverge permanently (one company gets into trouble). Mean reversion is a tendency, not a guarantee.

## 3.6 Jump Processes — When Smooth Models Break

### The Problem with Continuous Models

Brownian motion produces continuous paths. But financial prices can **gap** — jump discontinuously:

- A company announces terrible earnings after hours → stock opens 30% lower.
- A central bank unexpectedly cuts rates → bonds spike up.
- A geopolitical event → oil jumps 10% overnight.

These are not gradual moves that Brownian motion can capture.

### The Poisson Process — Modeling Arrivals

To model jumps, we need a process that captures **rare, discrete events**. The Poisson process does exactly this:

- Events arrive randomly with average rate $\lambda$ per year.
- The time between events is exponentially distributed.
- Events are independent of each other.

Think of it like lightning strikes: unpredictable timing, but you know the average frequency.

### Jump-Diffusion = Continuous + Discrete

Merton's model (1976) combines both:
- **Between jumps**: The stock follows standard GBM (smooth random walk).
- **At jump times**: The stock multiplies by a random factor (up or down instantaneously).

This creates paths that look like real stock prices: mostly smooth wiggles punctuated by occasional sharp moves.

### Impact on Options

Jump risk changes options pricing significantly:
- **OTM puts become more expensive**: The probability of a sudden crash is higher than GBM predicts.
- **The volatility smile appears naturally**: Jump models produce the U-shaped implied vol pattern without needing stochastic vol.
- **Hedging becomes harder**: You can't delta-hedge through a jump. If the stock jumps 10%, your hedge was based on the pre-jump delta — you lose money on the gap.

This unhedgeable jump risk is why option sellers demand a premium above Black-Scholes theoretical prices.

## 3.7 Quadratic Variation — The Hidden Quantity

### What It Is

Quadratic variation measures the **total accumulated squared movement** of a process. It's the sum of all the tiny squared increments.

### Why It Matters

Quadratic variation is the mathematical quantity that separates stochastic calculus from regular calculus.

For a smooth function (like $f(t) = \sin(t)$):
- Total variation (sum of |increments|): finite
- Quadratic variation (sum of increments²): **zero**

For Brownian motion:
- Total variation: **infinite** (paths are infinitely jagged)
- Quadratic variation: **finite and equals $T$**

The fact that $(dW)^2 = dt$ (a non-zero, deterministic quantity) is THE fundamental property that creates the Ito correction term and makes all of stochastic calculus different from regular calculus.

### Realized Variance

In practice, you can estimate quadratic variation from high-frequency data:

$$\text{Realized Variance} \approx \sum_{i} (\text{5-minute return}_i)^2$$

This is the basis for **realized volatility estimation** and **variance swaps** — derivatives that pay based on realized variance.

# 4. STATISTICAL TOOLS FOR FINANCE
---

## 4.1 PCA — Finding the Hidden Drivers

### The Problem

You have 500 stocks in your portfolio. Each has daily returns. That's 500 time series with complex inter-relationships. How do you make sense of this?

### What PCA Does

PCA finds a small number of **independent factors** that explain most of the movement in your data. It rotates your coordinate system to align with the directions of maximum variance.

### The Analogy

Imagine 500 puppets connected by strings to a few puppet masters:
- **Puppet Master 1** (the market): When they move, ALL puppets move. This is PC1 — typically explains 40-60% of all stock movement.
- **Puppet Master 2** (sector rotation): When they move, some puppets go up and others go down. This is PC2 — explains 10-15%.
- **Puppet Master 3** (value vs growth): Another independent source of movement. PC3 — explains 5-10%.
- Beyond PC5 or so: mostly individual puppet jiggling (noise).

### PCA in Yield Curves

PCA is especially clean in interest rate analysis:
- **PC1** ≈ level shift (all rates move up or down together) — 85% of variance
- **PC2** ≈ slope change (short rates and long rates move opposite) — 10% of variance  
- **PC3** ≈ curvature (middle rates move differently from ends) — 3% of variance
- Three numbers summarize the behavior of the entire yield curve!

### Dimensionality Reduction

If 5 components explain 90% of variance, you can replace 500 individual stock positions with exposure to 5 factors. This simplifies risk management enormously and reveals the **true sources of risk** hidden in a complex portfolio.

## 4.2 Kalman Filter — Tracking Hidden States

### The Problem

Many quantities in finance are **not directly observable** — they're hidden and must be inferred from noisy measurements:

- **True volatility**: We can compute realized vol, but it's a noisy estimate of the true underlying vol.
- **Fair value**: The "true" price of an asset is unobservable; we see bids, asks, and trades — all noisy.
- **Hedge ratios**: The true beta of a stock vs the market changes over time and can only be estimated.

### What the Kalman Filter Does

The Kalman filter is an algorithm that optimally estimates a hidden state from a sequence of noisy observations. It has two steps that repeat:

#### Step 1: Predict
"Based on what I knew yesterday and my model of how things evolve, here's my best guess for today."

#### Step 2: Update
"I just got a new measurement. Let me blend it with my prediction, weighting by how much I trust each."

### The Kalman Gain — The Trust Dial

The Kalman gain $K$ is a number between 0 and 1 that controls the blend:

- **$K$ close to 1**: "I trust the measurement more than my prediction" → the estimate closely follows new data (responsive but noisy).
- **$K$ close to 0**: "I trust my model more than the noisy measurement" → the estimate is smooth but slow to react.

$K$ is automatically computed based on the relative uncertainties. No manual tuning needed.

### Finance Applications

| Application | Hidden State | Measurement |
|-------------|--------------|-------------|
| Dynamic hedge ratio | True beta | Rolling regression estimate |
| Volatility filtering | True vol | Squared returns or realized vol |
| Market-making | True fair price | Bid-ask quotes, trades |
| Pairs trading | Spread equilibrium level | Observed spread |
| Signal extraction | True alpha signal | Noisy factor model residuals |

### Why Not Just Use Moving Averages?

Moving averages are a poor man's Kalman filter. The Kalman filter is better because:
1. It **adapts** the smoothing parameter (via Kalman gain) based on signal-to-noise ratio.
2. It provides **uncertainty estimates** — not just a point estimate but a confidence interval.
3. It's **optimal** — no other linear filter can produce a lower estimation error.

## 4.3 Copulas — Modeling How Things Break Together

### The Problem

You know the individual distribution of each stock's returns (the "marginals"). You know the correlation between them. But correlation doesn't tell the whole story:

- Do they crash together more often than they rally together? (**Asymmetric tail dependence**)
- Is the dependence constant, or does it change in extreme scenarios? (**Tail dependence**)
- Can you have zero correlation but strong dependence? (**Non-linear dependence**)

### Sklar's Theorem — The Breakthrough

Any joint distribution can be decomposed into:
1. **Marginals**: How each variable behaves individually.
2. **Copula**: How the variables are linked together.

This separation lets you model each part independently. You could have:
- Normally distributed stock returns with a Student-t copula (capturing tail dependence)
- Exponentially distributed loss amounts with a Clayton copula (capturing lower-tail dependence)

### The 2008 Connection — The Gaussian Copula Disaster

David Li's Gaussian copula model was widely used to price CDOs (Collateralized Debt Obligations).

The model assumed: **mortgage defaults are linked via a Gaussian copula**, which implies **no tail dependence** — defaults become independent in extreme scenarios.

Reality: when the housing market crashed, defaults became **highly correlated** — exactly the scenario the Gaussian copula said was nearly impossible. The model dramatically underestimated the probability of widespread simultaneous defaults.

Lesson: Copula choice matters enormously, especially in the tails. The Gaussian copula is fine for normal markets but dangerously wrong for crisis modeling.

### Types of Copulas

| Copula | Tail Dependence | Best For |
|--------|----------------|----------|
| **Gaussian** | None | Normal market conditions, simple models |
| **Student-t** | Both tails (symmetric) | Capturing co-crashes AND co-rallies |
| **Clayton** | Lower tail only | Assets that crash together but don't necessarily rally together |
| **Gumbel** | Upper tail only | Assets that rally together but crash independently |
| **Frank** | None (symmetric) | Weak, symmetric dependence |
