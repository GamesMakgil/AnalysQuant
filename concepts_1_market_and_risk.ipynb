# 1. VOLATILITY — The Language of Uncertainty
---

## 1.1 What Volatility Really Is

Volatility is the **heartbeat of financial markets**. It measures how much a price wiggles, jumps, and swings over time.

### The Core Idea

Imagine two stocks, both averaging +10% per year:
- **Stock A**: Climbs steadily, a little each day, almost like a staircase going up.
- **Stock B**: Rockets up 5% one day, crashes 4% the next, swings wildly but averages the same 10%.

Both have the same **return**, but Stock B has much higher **volatility**. Volatility captures **the journey, not just the destination**.

### Why Investors Care

1. **Risk perception**: High volatility feels dangerous, even if the average return is good. Humans experience losses more painfully than equivalent gains (loss aversion), so a volatile ride causes more psychological stress.

2. **Position sizing**: If you know a stock moves 2% per day on average, you can size your position so that a bad day doesn't wipe you out. If it moves 10% per day, you need a much smaller position for the same dollar risk.

3. **Options pricing**: Volatility is the single most important input in options pricing. Higher volatility = more expensive options, because the chance of the option ending in-the-money increases.

4. **Portfolio construction**: Combining low-correlation, moderate-volatility assets can create a portfolio with lower overall volatility than any individual holding — this is the essence of diversification.

## 1.2 Realized vs Implied Volatility — Past vs Future

This is one of the most important distinctions in all of finance.

### Realized Volatility (Historical)

- **What it is**: A statistical measurement computed from past price data.
- **Time direction**: Backward-looking — tells you what already happened.
- **Analogy**: Looking in the rearview mirror while driving. You can see the road you already traveled.
- **Strengths**: Objective, measurable, based on facts.
- **Weaknesses**: The past doesn't guarantee the future. A stock that was calm for 6 months can suddenly become volatile.

### Implied Volatility

- **What it is**: The market's collective forecast of future volatility, extracted from option prices.
- **Time direction**: Forward-looking — tells you what the market expects will happen.
- **Analogy**: Looking through the windshield. You can see what's coming (but not perfectly).
- **Strengths**: Incorporates the wisdom (and fear) of all market participants. Updates in real-time.
- **Weaknesses**: The market can be wrong. Implied vol includes a risk premium (insurance cost), so it systematically overestimates actual future vol.

### The Volatility Risk Premium (VRP)

On average, implied volatility is **higher** than realized volatility by about 2-4 percentage points. This gap is the **volatility risk premium** — the price of uncertainty insurance.

Think of it like car insurance: the premium you pay is higher than your expected accident cost. The insurance company profits from this gap. Similarly, option sellers collect the VRP as compensation for bearing volatility risk.

Many hedge fund strategies are built entirely around harvesting the VRP: sell options (collect implied vol premium), hedge the delta, and pocket the difference between what the market feared and what actually happened.

## 1.3 The Volatility Surface and Smile

### The Problem with Black-Scholes

Black-Scholes assumes volatility is **constant** — one number for all options on the same stock. But in reality, options with different strikes and expirations trade at **different implied volatilities**.

### The Volatility Smile

If you plot implied volatility against strike price for a single expiration, you get a U-shape or "smile":

- **Deep out-of-the-money puts** (low strikes): High implied vol — crash protection is expensive.
- **At-the-money options** (middle strikes): Lowest implied vol.
- **Deep out-of-the-money calls** (high strikes): Moderately high implied vol — upside lottery tickets.

### Why Does the Smile Exist?

1. **Fat tails**: Real returns are not normally distributed. Extreme moves (crashes, melt-ups) happen more often than the bell curve predicts. The smile compensates for this.

2. **Demand for protection**: Fund managers constantly buy downside puts for insurance. This demand pushes up the price (and thus implied vol) of OTM puts.

3. **Leverage effect**: When stocks drop, companies become more leveraged (debt/equity ratio rises), making them inherently riskier → higher vol. This creates a negative correlation between returns and vol.

### The Volatility Surface

Extend the smile across multiple expirations, and you get a 3D surface:
- X-axis: Strike price (or moneyness)
- Y-axis: Time to expiration
- Z-axis: Implied volatility

This surface is a living, breathing entity that changes with every tick. Understanding its dynamics is central to options market-making.

## 1.4 Forward Volatility — Reading Between the Lines

### The Concept

You know the implied vol for 1-year options and 2-year options. Can you figure out what the market expects vol to be **between year 1 and year 2**?

Yes — that's **forward volatility**.

### The Key Insight: Variance Is Additive, Volatility Is Not

This is crucial and counterintuitive:

- You **CANNOT** average volatilities: $(20\% + 30\%) / 2 \neq 25\%$ for the combined period.
- You **CAN** add variances: $\text{Var}_{0 \to 2} = \text{Var}_{0 \to 1} + \text{Var}_{1 \to 2}$.

This is because variance scales linearly with time (for independent increments), while standard deviation (volatility) scales with the square root of time.

### Why It Matters

- **Term structure trading**: If forward vol looks too high or low relative to your view, you can trade the difference (calendar spreads).
- **Event detection**: A spike in forward vol around a specific future date (e.g. an election) tells you the market is pricing in event risk.
- **Forward-starting options**: Some exotic derivatives start at a future date, and they need forward vol for pricing.

## 1.5 Compounding — The Eighth Wonder of the World

### Simple vs Compound Returns

**Simple return**: You earn interest on your original investment only.  
$1000 at 10% for 3 years → $1000 + $300 = $1300$

**Compound return**: You earn interest on your interest.  
$1000 at 10% for 3 years → $1000 \times 1.1^3 = $1331$

The $31 difference is interest earned on previous interest. Over longer periods, this snowball effect becomes massive.

### The Asymmetry of Returns

Compounding creates a fundamental asymmetry:

- **+50% then -50% ≠ 0%**: $(1.5)(0.5) = 0.75$ → you lost 25%!
- To recover from a -50% loss, you need a +100% gain.
- To recover from a -90% loss, you need a +900% gain.

This asymmetry is why **volatility drag** exists: a volatile portfolio underperforms a steady one with the same arithmetic average return.

### Log Returns vs Simple Returns

Quants often use **log returns** $r = \ln(P_t / P_{t-1})$ because:
1. Log returns are **additive** across time: $r_{total} = r_1 + r_2 + r_3$ (no need for multiplication).
2. Log returns are more **symmetric**: a +10% log return and -10% log return cancel perfectly.
3. Log returns naturally connect to **continuous compounding** and stochastic calculus.

# 2. RISK METRICS — Measuring What Can Go Wrong
---

## 2.1 Value at Risk (VaR) — The Industry Standard

### What It Is

VaR answers one question: **"What is the worst loss I can expect on a normal day?"**

More precisely: "With X% confidence, what is the maximum I will lose over the next N days?"

### Example

"The 1-day 95% VaR is $2 million" means:
- On 95% of trading days, losses will be less than $2M.
- On the remaining 5% of days (about 12.5 days per year), losses will exceed $2M.

### Three Methods of Computing VaR

#### 1. Parametric (Analytical)
- **Assumes** returns follow a normal distribution.
- **Pro**: Fast to compute, closed-form solution.
- **Con**: Real returns have fat tails — extreme events are more frequent than normal distribution predicts. This method underestimates tail risk.

#### 2. Historical Simulation
- **Assumes** the past is representative of the future.
- **Pro**: No distribution assumption needed. Captures fat tails, skewness, and complex dependencies that actually existed.
- **Con**: Limited by your data window. If you use 2 years of data, you only have ~500 observations. You will never see a "once in 100 years" event.

#### 3. Monte Carlo
- **Assumes** whatever model you specify (normal, t-distribution, stochastic vol, etc.).
- **Pro**: Most flexible. Can model complex portfolios with options, path-dependent payoffs, etc.
- **Con**: Computationally expensive. Only as good as the model you choose.

### The Fatal Flaw of VaR

VaR tells you the **boundary** of the loss region but nothing about what's **inside** it.

"95% VaR = $2M" means you lose more than $2M about 5% of the time. But *how much more?* $2.1M? $50M? VaR doesn't say.

This is like a weather forecast saying "there's a 5% chance of a storm" without telling you whether it's a drizzle or a hurricane.

## 2.2 Expected Shortfall (CVaR) — VaR's Smarter Cousin

### What It Is

Expected Shortfall (also called Conditional VaR or CVaR) answers: **"Given that I'm having a bad day (beyond VaR), how bad is it on average?"**

It is the **average loss in the tail** — the mean of all losses that exceed VaR.

### Why It's Better Than VaR

1. **Tail-aware**: It looks *inside* the danger zone, not just at the boundary.
2. **Coherent risk measure**: It satisfies all the mathematical properties a good risk measure should (notably, it respects diversification — the combined risk of two portfolios is ≤ sum of individual risks).
3. **Regulatory acceptance**: Basel III moved from VaR to Expected Shortfall for market risk capital.

### Intuition

If VaR is "the worst loss on a normal bad day", then CVaR is "the average loss on catastrophic days".

Example:
- 95% VaR = $2M
- 95% CVaR = $3.5M

This means: when you lose more than $2M (which happens ~5% of the time), the average loss on those days is $3.5M. Much more informative.

## 2.3 Sharpe Ratio — The Universal Scorecard

### What It Measures

The Sharpe ratio is the **return-per-unit-of-risk** metric that the entire investment industry uses to compare strategies, funds, and assets on equal footing.

### The Deep Intuition

Anyone can earn a high return by taking huge risk (just use 10x leverage on the S&P 500). The question isn't "how much did you make?" but **"how much did you make per unit of pain?"**

The Sharpe ratio normalizes returns by risk, creating a level playing field.

### Subtracting the Risk-Free Rate — Why?

If Treasury bills yield 5%, then any strategy that earns less than 5% with ANY risk is destroying value — you could do better with zero risk. The risk-free rate is the **opportunity cost** of taking risk.

By subtracting $R_f$, the Sharpe ratio only credits you for returns **above** what you could earn for free. This is the excess return, or "alpha" in its simplest form.

### Practical Benchmarks

- **S&P 500** (long-term): Sharpe ≈ 0.4-0.5
- **Good hedge fund**: Sharpe ≈ 1.0-1.5
- **Renaissance Medallion fund** (legendary): Sharpe ≈ 3-4 (nearly unheard of)
- **Individual stock-pickers** (average): Sharpe ≈ 0.1-0.3

### Limitations

1. **Symmetric risk penalty**: Treats upside and downside volatility equally. A strategy that has occasional big wins but steady small gains gets penalized.
2. **Time-dependent**: A daily Sharpe of 0.1 annualizes to $0.1 \times \sqrt{252} \approx 1.6$, which looks great. But are daily returns really independent? Serial correlation inflates the number.
3. **Manipulable**: Strategies that sell tail risk (e.g. selling deep OTM puts) can show artificially high Sharpe ratios — until they blow up.

## 2.4 Sortino Ratio — Rewarding the Right Kind of Risk

### The Problem with Sharpe

The Sharpe ratio treats ALL volatility as bad. But is upside volatility really bad? If your portfolio occasionally jumps up 20% in a month, that increases your volatility — and lowers your Sharpe ratio — even though it's exactly what you want.

### Sortino's Fix

The Sortino ratio replaces total volatility with **downside deviation** — it only counts the bad moves.

### When to Prefer Sortino

- **Asymmetric strategies**: Options buying (limited downside, unlimited upside), trend following, venture capital.
- **Skewed returns**: Any strategy where the return distribution is not symmetric.
- **Behavioral alignment**: It matches what investors actually care about — losses, not volatility per se.

### Example

Two strategies with identical Sharpe = 1.0:
- **Strategy A**: Steady returns, occasionally a big loss → Sortino = 0.8 (downside vol is significant)
- **Strategy B**: Steady returns, occasionally a big win → Sortino = 2.5 (little downside vol)

The Sharpe ratio says they're equal. The Sortino ratio correctly identifies B as vastly superior for a risk-averse investor.

## 2.5 RAROC — The Banker's Decision Tool

### Context: Why Banks Need It

Banks are complex organizations with dozens of business lines:
- Mortgage lending
- Corporate loans
- Derivatives trading
- Wealth management
- Credit cards

Each generates different revenues and consumes different amounts of **economic capital** (the buffer against unexpected losses). The CEO needs to know: **which businesses are creating the most value per unit of capital consumed?**

### Economic Capital vs Regulatory Capital

- **Regulatory capital**: The minimum capital a bank must hold by law (Basel rules). A compliance floor.
- **Economic capital**: The capital the bank's own risk models say it needs to survive extreme losses with high confidence (e.g. 99.9%). Usually higher than regulatory capital. This is the realistic risk buffer.

### How RAROC Drives Decisions

The bank sets a **hurdle rate** (minimum acceptable RAROC), typically the cost of equity (8-12%).

- **RAROC > hurdle rate**: The business unit creates shareholder value → allocate more capital.
- **RAROC < hurdle rate**: The business unit destroys value → shrink it, reprice products, or exit.
- **RAROC = hurdle rate**: Break-even → look for ways to improve efficiency.

### The Power of RAROC

Without RAROC, a bank might celebrate a trading desk earning $100M — until you realize it needed $2B of capital to do so (5% RAROC, below the 10% hurdle). That capital could have earned more elsewhere.

RAROC forces every business to justify its capital consumption with sufficient risk-adjusted returns.

# 3. PORTFOLIO THEORY — Diversification and Beyond
---

## 3.1 Diversification — The Only Free Lunch in Finance

### The Core Theorem

If you combine assets that don't move perfectly together (correlation < 1), the portfolio's volatility is **less than** the weighted average of individual volatilities.

This is genuinely remarkable: you can reduce risk without reducing expected return.

### Why It Works

When Stock A has a bad day and Stock B has a good day, they partially cancel out. The more often this happens (lower correlation), the more the portfolio smooths out.

### The Limits of Diversification

1. **Correlation is not constant**: In crashes, correlations spike toward 1.0. Exactly when you need diversification most, it breaks down. This is called "correlation breakdown in crises."

2. **Diminishing returns**: Going from 1 stock to 10 stocks dramatically reduces risk. Going from 100 to 110 stocks does almost nothing. Most diversification benefit is captured within 20-30 uncorrelated positions.

3. **Systematic risk**: Even a perfectly diversified stock portfolio is exposed to market-wide risk (recessions, interest rate changes, pandemics). You can diversify away **idiosyncratic risk** (company-specific) but not **systematic risk** (market-wide).

4. **Over-diversification**: Too many positions can dilute your best ideas and increase transaction costs. Warren Buffett: "Diversification is protection against ignorance. It makes little sense if you know what you're doing."

## 3.2 Correlation vs Dependence

### Correlation Is Not the Whole Story

Pearson correlation measures **linear** dependence between two variables. But dependence can be:

- **Non-linear**: Two variables might be strongly related, but not in a straight line. Correlation can be zero even with perfect non-linear dependence.
- **Tail-dependent**: Two stocks might have low overall correlation but crash together. Correlation captures the average relationship, not the extreme behavior.

### Real-World Example

During 2006-2007, mortgage-backed securities had "low correlations" with each other. Risk models said: "diversified, safe!"

In 2008, they ALL crashed simultaneously. The average correlation was low, but the **tail dependence** was extreme. Everyone was exposed to the same underlying risk (housing prices), and when that broke, everything broke together.

This is why copulas (covered in the next notebook) were invented — to model dependence beyond simple correlation.

## 3.3 Fat Tails — Why Normal Isn't Normal

### The Problem

Most quantitative finance assumes returns follow a **normal (Gaussian) distribution**. Under this assumption:
- A 3-sigma event (3 standard deviations) happens about 0.3% of the time.
- A 5-sigma event happens about 0.00006% of the time (once in 14,000 years).
- A 10-sigma event is essentially impossible ($10^{-23}$ probability).

### The Reality

- **Black Monday (1987)**: The S&P 500 dropped 20.5% in one day — a 20+ sigma event. Under normality, this should never happen in the history of the universe.
- **2008 Financial Crisis**: Multiple 5-10 sigma daily moves.
- **Flash Crash (2010)**: Dow dropped ~9% in minutes.
- **March 2020**: Multiple days with 7-10% S&P swings.

### Why Do Fat Tails Exist?

1. **Feedback loops**: Panic selling triggers more panic selling. Stop-losses cascade. Margin calls force liquidations.
2. **Leverage**: When leveraged positions unwind, the forced selling amplifies moves.
3. **Information clustering**: Major news (war, pandemic, bankruptcy) doesn't arrive smoothly — it hits all at once.
4. **Behavioral herding**: Humans follow crowds, creating momentum beyond what fundamentals justify.

### Consequences for Risk Management

- **VaR underestimates tail risk** when using normal distributions.
- **Options are underpriced** if you assume normality (the true probability of extreme moves is higher).
- **Diversification fails** when you need it most (correlations spike in tails).
- Smart risk managers use **Student-t distributions**, **extreme value theory**, or **historical simulation** to capture fat tails.

## 3.4 Annualization — The Time-Scaling Trap

### Why We Annualize

Returns and volatility are computed over various time horizons (daily, weekly, monthly). To compare apples-to-apples, we convert everything to an annual basis.

### The Square Root of Time Rule

If daily returns are independent and identically distributed:
- **Variance** scales linearly with time: $\sigma^2_{annual} = 252 \times \sigma^2_{daily}$
- **Volatility** scales with the square root: $\sigma_{annual} = \sqrt{252} \times \sigma_{daily} \approx 15.87 \times \sigma_{daily}$
- **Returns** scale linearly: $R_{annual} = 252 \times R_{daily}$ (arithmetic) or compound

### When the Rule Breaks

1. **Serial correlation**: If today's return predicts tomorrow's (momentum or mean-reversion), the i.i.d. assumption fails.
   - Positive autocorrelation → true annual vol is HIGHER than $\sqrt{252} \times \sigma_{daily}$
   - Negative autocorrelation → true annual vol is LOWER

2. **Volatility clustering**: Calm periods and turbulent periods alternate (the GARCH effect). This violates the "identically distributed" part.

3. **Leverage and rebalancing**: Daily-rebalanced leveraged ETFs can have very different annualized behavior than $\sqrt{252}$ scaling would suggest.

### The 252 Number

Why 252? It's the approximate number of trading days in a year (Mon-Fri, minus holidays). For calendar days use 365. For weeks use 52. The choice matters and should be consistent with how your data is sampled.

# 4. MARKET MICROSTRUCTURE — How Trading Actually Works
---

## 4.1 The Order Book

### What It Is

The order book is the **living registry** of all outstanding buy and sell orders for a security. It's the mechanism through which price discovery happens.

### Structure

```
SELL (Ask) Side              BUY (Bid) Side
$101.50  ×  200 shares      $100.50  ×  500 shares  ← Best Bid
$101.25  ×  300 shares      $100.25  ×  800 shares
$101.00  ×  1000 shares     $100.00  ×  1200 shares
   ↑ Best Ask
```

- **Best bid**: Highest price someone is willing to pay ($100.50)
- **Best ask**: Lowest price someone is willing to sell ($101.00)
- **Spread**: $101.00 - $100.50 = $0.50 — the cost of immediacy

### Why the Spread Exists

Market makers provide liquidity by posting both bids and asks. The spread is their compensation for:
1. **Adverse selection**: Trading against informed traders who know the price is about to move.
2. **Inventory risk**: Holding a position that might lose value.
3. **Operational costs**: Technology, compliance, capital.

Spreads are tighter for liquid assets (major stocks, EUR/USD) and wider for illiquid ones (small caps, exotic derivatives).

## 4.2 Slippage and Price Impact

### Slippage — The Hidden Transaction Cost

When you see a stock quoted at $100 and you buy, you might actually pay $100.05. That $0.05 gap is **slippage**.

Slippage arises because:
1. **The quote changed** between your decision and your order reaching the exchange (latency).
2. **You consumed liquidity** at the best price and had to dip into worse prices (market depth).
3. **Other traders saw your order** and adjusted their quotes (information leakage).

### Price Impact — You Move the Market

When you place a large order, YOU change the price. This is price impact.

Imagine the order book has 1,000 shares available at $100, 1,000 at $100.05, and 1,000 at $100.10. If you buy 2,500 shares:
- First 1,000 fill at $100
- Next 1,000 fill at $100.05
- Last 500 fill at $100.10
- Average fill: $100.03 (vs the initial $100)

Your order literally walked up the order book, consuming liquidity at each level.

### Why Execution Matters

For a quant fund trading $50M/day:
- 10 bps of slippage = $50,000/day = $12.5M/year
- This can be the difference between a profitable and losing strategy
- Execution optimization (TWAP, VWAP, smart order routing) is an entire discipline

## 4.3 TWAP and VWAP — The Algorithms Behind the Trades

### The Problem

You need to buy 100,000 shares of AAPL. If you submit one big market order, you'll massively impact the price and everyone will see what you're doing. Solution: break the order into small pieces.

### TWAP (Time-Weighted Average Price)

**Strategy**: Execute equal amounts at regular intervals.

100,000 shares over 6.5 hours → ~256 shares every minute.

**Pros**:
- Dead simple to implement.
- Predictable execution schedule.
- Works well when you have no view on intraday volume patterns.

**Cons**:
- Ignores market conditions — trades the same amount during quiet and active periods.
- Predictable schedule makes you vulnerable to gaming by other algorithms.

### VWAP (Volume-Weighted Average Price)

**Strategy**: Execute more when the market is busiest, less when it's quiet.

Market volume is typically U-shaped during the day (high at open, low at lunch, high at close). VWAP follows this pattern.

**Pros**:
- Mimics the market's natural rhythm — less price impact.
- Industry-standard benchmark for institutional execution quality.
- Harder to game because you're trading with the crowd, not against it.

**Cons**:
- Requires a forecast of the volume profile (usually based on historical averages).
- If actual volume deviates from the forecast, execution quality suffers.
- Still predictable enough that sophisticated opponents can exploit it.

# 5. CREDIT RISK — The Risk of Not Getting Paid
---

## 5.1 Default — When Promises Break

### What Is Default?

Default occurs when a borrower **fails to meet their contractual obligations** — typically failing to make an interest or principal payment on time.

### The Three Pillars of Credit Loss

#### Probability of Default (PD)

How likely is this borrower to default? Estimated using:
- **Credit ratings**: S&P, Moody's, Fitch (AAA → D)
- **Financial ratios**: Debt/equity, interest coverage, cash flow
- **Market signals**: CDS spreads, bond yields, stock price
- **Statistical models**: Logistic regression, machine learning on historical defaults

Historical 1-year PDs by rating (rough averages):
| Rating | PD |
|--------|------|
| AAA | 0.01% |
| AA | 0.02% |
| A | 0.05% |
| BBB | 0.20% |
| BB | 1.0% |
| B | 4.0% |
| CCC | 15% |

#### Loss Given Default (LGD)

If default happens, how much do you lose? Not all defaults are 100% losses:
- **Secured loans** (backed by collateral): LGD ≈ 20-40%. You sell the collateral and recover some money.
- **Unsecured bonds**: LGD ≈ 50-70%. You wait in line during bankruptcy proceedings.
- **Equity-like instruments**: LGD ≈ 100%. Last in line, usually get nothing.

The complement, **Recovery Rate = 1 - LGD**, is how much you get back.

#### Exposure at Default (EAD)

How much money is at stake when default occurs?
- For a fixed loan: EAD = outstanding balance (straightforward).
- For a credit line: EAD = drawn amount + estimated future drawdown. Borrowers in distress tend to draw down their credit lines before defaulting — so EAD is usually higher than current balance.

## 5.2 Loan-to-Value (LTV) — The Collateral Buffer

### The Concept

LTV measures how leveraged a borrower is relative to their collateral. Think of it as the answer to: **"How much of the house did you buy with borrowed money?"**

### The Equity Cushion

If LTV = 80%, the borrower has 20% equity ("skin in the game"). The collateral must lose more than 20% of its value before the loan is "underwater" (owed more than it's worth).

This cushion protects the lender in two ways:
1. **Economic protection**: Even if the borrower defaults, selling the collateral covers the loan.
2. **Behavioral incentive**: The borrower has equity to protect, so they're more motivated to keep paying. When LTV exceeds 100% (underwater), the borrower has an incentive to walk away — this was a major driver of the 2008 mortgage crisis.

### LTV in Different Contexts

| Context | Typical LTV | Collateral |
|---------|-------------|------------|
| Prime mortgages | 70-80% | House |
| Subprime mortgages (pre-2008) | 95-100% | House |
| Margin loans (brokerage) | 50% | Stock portfolio |
| DeFi lending (Aave) | 50-75% | Cryptocurrency |
| Lombard loans (private banking) | 50-70% | Investment portfolio |

### Dynamic LTV

LTV changes over time as both the loan balance and collateral value change:
- **Amortization** (loan payments) → LTV decreases.
- **Collateral appreciation** → LTV decreases.
- **Collateral depreciation** → LTV increases → danger zone.

This is why DeFi protocols monitor LTV (via Health Factor) continuously and liquidate positions automatically when LTV gets too high.

## 5.3 Expected vs Unexpected Loss

### Expected Loss (EL) — The Cost of Doing Business

Every loan portfolio will experience some defaults. EL is the **average loss** you anticipate over the life of the portfolio. It's like shrinkage in retail — a predictable cost.

Banks handle EL through **loan loss provisions** — money set aside from revenue to cover anticipated losses. This is baked into the pricing: the interest rate on a risky loan is higher specifically to cover the expected losses.

### Unexpected Loss (UL) — The Surprise

The actual loss in any given year can be much higher than the average. UL measures the **variability** of losses around the expected value.

Banks handle UL through **economic capital** — a safety buffer of equity that absorbs losses beyond what provisions cover.

### Stress Loss — The Catastrophe

In a severe recession, losses can far exceed even the capital buffer. Banks handle this through:
- **Stress testing**: "What if GDP drops 10%? What if unemployment hits 15%?"
- **Government backstops**: Central bank lending, deposit insurance
- **Too big to fail**: Systemic risk regulation

### The Hierarchy

```
Losses covered by:     provisions  →  economic capital  →  government/default
Probability:           high            moderate              very low
Type:                  Expected        Unexpected            Stress/Tail
```
